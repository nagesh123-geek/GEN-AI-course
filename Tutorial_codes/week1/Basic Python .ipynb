# Functions 

```python
def add_numbers(a,b):
    result = a + b
    return result
```

```python
answer = add_numbers(10,20)
print("Answer :", answer)
```

```python
# Compute mean 

def compute_mean(values):
    total = 0
    for value in values:
        total = total + value
    mean = total / len(values)
    return mean
```

```python
marks = [70,80,90,70,80]
mean_marks = compute_mean(marks)
print("Mean Marks:", mean_marks)
```

# Intro to OOP(Object Oriented Programming)

# 1. CLASSES

```python
class student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks
        
    def display(self):
        print("Student Name :", self.name)
        print("Marks :", self.marks)
        
```

```python
# Student 1 : Instances of classes aka objects

student1 = student("Nagesh",75)
```

```python
student1.display()
```

# Simple Linear Model class 

```python
class Simple_Linear_Model:
    def __init__(self,weight,bias):
        self.weight = weight
        self.bias = bias
        
    def forward(self,x):
        y = self.weight * x + self.bias
        return y

        
```

```python
model = Simple_Linear_Model(weight = 2, bias = 5)
```
