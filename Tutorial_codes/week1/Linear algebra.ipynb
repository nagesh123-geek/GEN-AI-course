# Reflection in 2d 

```python
import numpy as np
import matplotlib.pyplot as plt

# Original vector
v = np.array([2, 1])

# Reflection about x-axis
R = np.array([[1, 0],
              [0,-1]])

v_ref = R @ v

plt.figure(figsize=(6,6))

plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='blue',
           label='Original')

plt.quiver(0,0,v_ref[0],v_ref[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='red',
           label='Reflection')

plt.xlim(-3,3)
plt.ylim(-3,3)
plt.grid()
plt.axis('equal')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rect = np.array([
    [0,0],
    [4,0],
    [4,2],
    [0,2],
    [0,0]
]).T

R = np.array([
    [1,0],
    [0,-1]
])

reflected = R @ rect

plt.figure(figsize=(6,6))

plt.plot(rect[0], rect[1], lw=3, label="Original")
plt.plot(reflected[0], reflected[1], lw=3, label="Reflection")

plt.axis("equal")
plt.grid(True)
plt.xlim(-1,5)
plt.ylim(-3,3)
plt.legend()
plt.title("Reflection about x-axis")

plt.show()
```

# Rotation in 2d 

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.radians(45)

R = np.array([[np.cos(theta),-np.sin(theta)],
              [np.sin(theta), np.cos(theta)]])

v = np.array([2,1])

v_rot = R @ v

plt.figure(figsize=(6,6))

plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='blue')

plt.quiver(0,0,v_rot[0],v_rot[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='red')

plt.grid()
plt.axis('equal')
plt.xlim(-3,3)
plt.ylim(-3,3)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Rectangle vertices
rect = np.array([
    [0, 0],
    [4, 0],
    [4, 2],
    [0, 2],
    [0, 0]
]).T

# Rotation angle
theta = np.radians(40)

R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

rotated = R @ rect

plt.figure(figsize=(6,6))

plt.plot(rect[0], rect[1], 'b-', lw=3, label="Original")
plt.plot(rotated[0], rotated[1], 'r-', lw=3, label="Rotated")

plt.axis("equal")
plt.grid(True)
plt.xlim(-3,5)
plt.ylim(-2,5)
plt.legend()
plt.title("2D Rotation")
plt.show()
```

# Rotation in 3d 

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

# Rectangle in the xy-plane
rect = np.array([
    [0,0,0],
    [4,0,0],
    [4,2,0],
    [0,2,0],
    [0,0,0]
]).T

# Rotation about z-axis
theta = np.radians(45)

R = np.array([
    [np.cos(theta),-np.sin(theta),0],
    [np.sin(theta), np.cos(theta),0],
    [0,0,1]
])

rotated = R @ rect

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111,projection='3d')

ax.plot(rect[0],rect[1],rect[2],
        lw=3,label='Original')

ax.plot(rotated[0],rotated[1],rotated[2],
        lw=3,label='Rotated')

ax.set_xlim(-3,5)
ax.set_ylim(-3,5)
ax.set_zlim(-1,3)

ax.legend()

plt.show()
```

# Reflection in 3d 

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

# -------------------------------------------------
# Rectangle in 3D
# -------------------------------------------------

rect = np.array([
    [0,0,1],
    [4,0,1],
    [4,2,1],
    [0,2,1],
    [0,0,1]
]).T

# -------------------------------------------------
# Reflection across the xy-plane
# -------------------------------------------------

R = np.array([
    [1,0,0],
    [0,1,0],
    [0,0,-1]
])

reflected = R @ rect

# -------------------------------------------------
# Plot
# -------------------------------------------------

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(rect[0], rect[1], rect[2],
        linewidth=3,
        label="Original")

ax.plot(reflected[0], reflected[1], reflected[2],
        linewidth=3,
        label="Reflected")

ax.set_xlim(-1,5)
ax.set_ylim(-1,3)
ax.set_zlim(-2,2)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.legend()
plt.title("Reflection across XY Plane")
plt.show()
```

# Non linear transformations are not possible ....

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------
# Create grid
# ------------------------------------------------
x = np.linspace(-3, 3, 21)
y = np.linspace(-3, 3, 21)
X, Y = np.meshgrid(x, y)

# ------------------------------------------------
# Change this matrix and try anything you like!
# Rotation, shear, scaling, reflection...
# ------------------------------------------------

A = np.array([[1.5, 0.8],
              [0.3, 1.2]])

# Apply linear transformation
X2 = A[0,0]*X + A[0,1]*Y
Y2 = A[1,0]*X + A[1,1]*Y

fig, ax = plt.subplots(figsize=(7,7))

# Horizontal lines
for i in range(X.shape[0]):
    ax.plot(X2[i,:], Y2[i,:], 'b')

# Vertical lines
for j in range(X.shape[1]):
    ax.plot(X2[:,j], Y2[:,j], 'b')

ax.set_aspect('equal')
ax.set_title("Any Matrix Transformation")
ax.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3,3,21)
y = np.linspace(-3,3,21)
X,Y = np.meshgrid(x,y)

A=np.array([[1.3,0.4],
            [0.2,1.1]])

B=np.array([[0.7,-0.5],
            [0.8,1.2]])

C=np.array([[1.1,0.6],
            [-0.3,0.9]])

# Combined matrix
M=C@B@A

X2=M[0,0]*X+M[0,1]*Y
Y2=M[1,0]*X+M[1,1]*Y

fig,ax=plt.subplots(figsize=(7,7))

for i in range(X.shape[0]):
    ax.plot(X2[i],Y2[i])

for j in range(X.shape[1]):
    ax.plot(X2[:,j],Y2[:,j])

ax.set_title("Three Linear Layers = One Matrix")
ax.set_aspect('equal')
ax.grid(True)
plt.show()
```

# With a non linear function (ReLU)

```python
import numpy as np
import matplotlib.pyplot as plt

def relu(z):
    return np.maximum(0,z)

x=np.linspace(-3,3,21)
y=np.linspace(-3,3,21)
X,Y=np.meshgrid(x,y)

A=np.array([[1.5,0.8],
            [0.3,1.2]])

# Linear layer
X1=A[0,0]*X+A[0,1]*Y
Y1=A[1,0]*X+A[1,1]*Y

# ReLU activation
X2=relu(X1)
Y2=relu(Y1)

fig,ax=plt.subplots(figsize=(7,7))

for i in range(X.shape[0]):
    ax.plot(X2[i],Y2[i],'r')

for j in range(X.shape[1]):
    ax.plot(X2[:,j],Y2[:,j],'r')

ax.set_title("Matrix + ReLU")
ax.set_aspect('equal')
ax.grid(True)
plt.show()
```

# Circle to banana using non linearity 

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# Original circle
# ==========================================================
theta = np.linspace(0, 2*np.pi, 500)

x = np.cos(theta)
y = np.sin(theta)

# ==========================================================
# Linear transformation (Matrix)
# ==========================================================
A = np.array([[1.5, 0.7],
              [0.3, 1.2]])

linear = A @ np.vstack([x, y])

# ==========================================================
# Nonlinear transformation
#
# This bends the ellipse into a "banana".
#
# x' = x
# y' = y + alpha*x^2
#
# The x² term is nonlinear.
# ==========================================================
alpha = 0.8

xb = linear[0]
yb = linear[1] + alpha * linear[0]**2

# ==========================================================
# Plot
# ==========================================================
fig, ax = plt.subplots(1, 3, figsize=(15,5))

# ----------------------------------------------------------
# Original
# ----------------------------------------------------------
ax[0].plot(x, y, lw=3)
ax[0].set_title("Original Circle")
ax[0].set_aspect('equal')
ax[0].grid(True)

# ----------------------------------------------------------
# Matrix
# ----------------------------------------------------------
ax[1].plot(linear[0], linear[1], lw=3, color='blue')
ax[1].set_title("After Matrix\n(Always an Ellipse)")
ax[1].set_aspect('equal')
ax[1].grid(True)

# ----------------------------------------------------------
# Nonlinear
# ----------------------------------------------------------
ax[2].plot(xb, yb, lw=3, color='red')
ax[2].set_title("Matrix + Nonlinearity\n(Banana Shape)")
ax[2].set_aspect('equal')
ax[2].grid(True)

plt.tight_layout()
plt.show()
```

# Nueral network style 

```python
import numpy as np
import matplotlib.pyplot as plt

# Circle
theta = np.linspace(0, 2*np.pi, 600)
X = np.vstack([np.cos(theta), np.sin(theta)])

# First linear layer
W1 = np.array([[2.0, 0.5],
               [0.3, 1.6]])

Z = W1 @ X

# Nonlinear activation
A = np.tanh(Z)

# Second linear layer
W2 = np.array([[1.2, 1.0],
               [-0.8, 1.4]])

Y = W2 @ A

plt.figure(figsize=(12,4))

plt.subplot(131)
plt.plot(X[0], X[1], lw=3)
plt.title("Circle")
plt.axis("equal")
plt.grid(True)

plt.subplot(132)
plt.plot(Z[0], Z[1], lw=3)
plt.title("After Linear Layer")
plt.axis("equal")
plt.grid(True)

plt.subplot(133)
plt.plot(Y[0], Y[1], lw=3, color="red")
plt.title("Linear → tanh → Linear")
plt.axis("equal")
plt.grid(True)

plt.tight_layout()
plt.show()
```

# Random matrix of circle .....

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# ==========================================================
# Generate random points uniformly inside a circle
# ==========================================================
N = 5000

theta = np.random.uniform(0, 2*np.pi, N)
r = np.sqrt(np.random.uniform(0, 1, N))

x = r * np.cos(theta)
y = r * np.sin(theta)

X = np.vstack([x, y])

# ==========================================================
# Linear layer
# ==========================================================
W = np.array([[1.6, 0.7],
              [0.3, 1.2]])

Z = W @ X

# ==========================================================
# ReLU activation
# ==========================================================
A = np.maximum(0, Z)

# ==========================================================
# Second linear layer
# ==========================================================
V = np.array([[1.2, -0.8],
              [0.5,  1.4]])

Y = V @ A

# ==========================================================
# Plot
# ==========================================================
fig, ax = plt.subplots(1,4,figsize=(10,5))

# ----------------------------------------------------------
# Original
# ----------------------------------------------------------
ax[0].scatter(X[0], X[1], s=3)
ax[0].set_title("Original Circle")

# ----------------------------------------------------------
# Matrix
# ----------------------------------------------------------
ax[1].scatter(Z[0], Z[1], s=3,color='blue')
ax[1].set_title("After Matrix\n(Ellipse)")

# ----------------------------------------------------------
# ReLU
# ----------------------------------------------------------
ax[2].scatter(A[0], A[1], s=3,color='red')
ax[2].set_title("After ReLU\n(Space Folded)")

# ----------------------------------------------------------
# Second Matrix
# ----------------------------------------------------------
ax[3].scatter(Y[0], Y[1], s=3,color='green')
ax[3].set_title("Linear → ReLU → Linear")

for a in ax:
    a.set_aspect('equal')
    a.grid(alpha=0.3)
    a.set_xlim(-3,3)
    a.set_ylim(-3,3)

plt.tight_layout()
plt.show()
```

# Random matrix to banana ....

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# ==========================================================
# Generate random points uniformly inside a circle
# ==========================================================
N = 5000

theta = np.random.uniform(0, 2*np.pi, N)
r = np.sqrt(np.random.uniform(0, 1, N))

x = r * np.cos(theta)
y = r * np.sin(theta)

X = np.vstack([x, y])

# ==========================================================
# Linear layer
# ==========================================================
W = np.array([[1.6, 0.7],
              [0.3, 1.2]])

Z = W @ X



x = Z[0]
y = Z[1]

# Nonlinear bending
alpha = 0.8

xb = x
yb = y + alpha*x**2

Y = np.vstack([xb,yb])


# ==========================================================
# ReLU activation
# ==========================================================
A = np.maximum(0, Z)

# ==========================================================
# Second linear layer
# ==========================================================
V = np.array([[1.2, -0.8],
              [0.5,  1.4]])

YR = V @ A



# ==========================================================
# Plot
# ==========================================================
fig, ax = plt.subplots(1,4,figsize=(10,5))

# ----------------------------------------------------------
# Original
# ----------------------------------------------------------
ax[0].scatter(X[0], X[1], s=3)
ax[0].set_title("Original Circle")

# ----------------------------------------------------------
# Matrix
# ----------------------------------------------------------
ax[1].scatter(Z[0], Z[1], s=3,color='blue')
ax[1].set_title("After Matrix\n(Ellipse)")

# ----------------------------------------------------------
# ReLU
# ----------------------------------------------------------
ax[2].scatter(YR[0], YR[1], s=3,color='red')
ax[2].set_title("After ReLU\n(Space Folded)")

# ----------------------------------------------------------
# Second Matrix
# ----------------------------------------------------------
ax[3].scatter(Y[0], Y[1], s=3,color='green')
ax[3].set_title("Linear → non Linear")

for a in ax:
    a.set_aspect('equal')
    a.grid(alpha=0.3)
    a.set_xlim(-3,3)
    a.set_ylim(-3,3)

plt.tight_layout()
plt.show()
```
