# ReLU activation 

```python
def relu(x):
    return np.maximum(0,x)
```

```python
x = np.array([-3,-1,0,2,5])
y = relu(x)
print(y)
```

```python
import matplotlib.pyplot as plt

plt.plot(x, y, marker='o')
plt.xlabel("x")
plt.ylabel("ReLU(x)")
plt.title("ReLU Activation Function")
plt.grid(True)
plt.show()
```

# Sigmoid Activation 

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
x = np.array([-9,-7,-5,-3,-1,0,1,3,5,7,9])
y = sigmoid(x)
print(y)
```

```python
import matplotlib.pyplot as plt

plt.plot(x, y, marker='o')
plt.xlabel("x")
plt.ylabel("Sigmoid(x)")
plt.title("Sigmoid Activation Function")
plt.grid(True)
plt.show()
```

# Softmax

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / (np.sum(exp_x))
```

```python
x = np.array([-9,-7,-5,-3,-1,0,1,3,5,7,9])
y = softmax(x)
print(y)
```

```python
import matplotlib.pyplot as plt

plt.plot(x, y, marker='o')
plt.xlabel("x")
plt.ylabel("Softmax(x)")
plt.title("Softmax Activation Function")
plt.grid(True)
plt.show()
```

# Manual Forward Pass

# Input --> Linear --> ReLU --> Linear ---> Softmax

```python
np.random.seed(42)
```

```python
# One input sample with 4 features
x = np.array([[1.0,2.0,3.0,4.0]])
```

```python
# First Layer
# 4 features --> 5 hidden neurons

W1 = np.random.randn(4,5)
b1 = np.random.randn(5)

# Second Laer 
# 5 hidden --> 3 classes

W2 = np.random.randn(5,3)
b2 = np.random.randn(3)
```

```python
# Forward Pass

z1 = x @ W1 + b1
a1 = relu(z1)

z2 = a1 @ W2 + b2

output = softmax(z2.flatten())

pred = np.argmax(output)
```

# MSE Loss

```python
def mse_loss(y_true,y_pred):
    return np.mean((y_true - y_pred)**2)
```

```python
y_true = np.array([10,20,30])
y_pred = np.array([12,18,33])

loss = mse_loss(y_true,y_pred)
print(loss)
```

```python
x = np.arange(len(y_true))

plt.plot(x, y_true, marker='o', label="True Values")
plt.plot(x, y_pred, marker='o', label="Predicted Values")

plt.xlabel("Sample")
plt.ylabel("Value")
plt.title(f"True vs Predicted Values (MSE = {loss:.2f})")
plt.legend()
plt.grid(True)
plt.show()
```

# Cross Entropy Loss

```python
def cross_entropy_loss(probabilities,true_class):
    return -np.log(probabilities[true_class])
```

```python
probabilities = np.array([0.1,0.7,0.3])
true_class = 1
loss = cross_entropy_loss(probabilities,true_class)
print(loss)
```

```python
classes = np.arange(len(probabilities))

plt.bar(classes, probabilities)

# Highlight true class
plt.bar(true_class, probabilities[true_class])

plt.xticks(classes, [f"Class {i}" for i in classes])
plt.xlabel("Classes")
plt.ylabel("Probability")
plt.title(f"Class Probabilities (Cross-Entropy Loss = {loss:.4f})")

plt.show()
```
