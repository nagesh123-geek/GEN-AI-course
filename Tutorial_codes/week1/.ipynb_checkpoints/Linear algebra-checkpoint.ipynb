# Reflection in 2d 

```python
import numpy as np
import matplotlib.pyplot as plt

# Original vector
v = np.array([2, 1])

# Reflection about x-axis
R = np.array([[1, 0],
              [0,-1]])

v_ref = R @ v

plt.figure(figsize=(6,6))

plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='blue',
           label='Original')

plt.quiver(0,0,v_ref[0],v_ref[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='red',
           label='Reflection')

plt.xlim(-3,3)
plt.ylim(-3,3)
plt.grid()
plt.axis('equal')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rect = np.array([
    [0,0],
    [4,0],
    [4,2],
    [0,2],
    [0,0]
]).T

R = np.array([
    [1,0],
    [0,-1]
])

reflected = R @ rect

plt.figure(figsize=(6,6))

plt.plot(rect[0], rect[1], lw=3, label="Original")
plt.plot(reflected[0], reflected[1], lw=3, label="Reflection")

plt.axis("equal")
plt.grid(True)
plt.xlim(-1,5)
plt.ylim(-3,3)
plt.legend()
plt.title("Reflection about x-axis")

plt.show()
```

# Rotation in 2d 

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.radians(45)

R = np.array([[np.cos(theta),-np.sin(theta)],
              [np.sin(theta), np.cos(theta)]])

v = np.array([2,1])

v_rot = R @ v

plt.figure(figsize=(6,6))

plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='blue')

plt.quiver(0,0,v_rot[0],v_rot[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='red')

plt.grid()
plt.axis('equal')
plt.xlim(-3,3)
plt.ylim(-3,3)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Rectangle vertices
rect = np.array([
    [0, 0],
    [4, 0],
    [4, 2],
    [0, 2],
    [0, 0]
]).T

# Rotation angle
theta = np.radians(40)

R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

rotated = R @ rect

plt.figure(figsize=(6,6))

plt.plot(rect[0], rect[1], 'b-', lw=3, label="Original")
plt.plot(rotated[0], rotated[1], 'r-', lw=3, label="Rotated")

plt.axis("equal")
plt.grid(True)
plt.xlim(-3,5)
plt.ylim(-2,5)
plt.legend()
plt.title("2D Rotation")
plt.show()
```

# Shear in 2D

```python
import numpy as np
import matplotlib.pyplot as plt

S = np.array([[1,1],
              [0,1]])

v = np.array([2,1])

v2 = S @ v

plt.figure(figsize=(6,6))

plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='blue')

plt.quiver(0,0,v2[0],v2[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='red')

plt.grid()
plt.axis('equal')
plt.xlim(-1,5)
plt.ylim(-1,4)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rect = np.array([
    [0,0],
    [4,0],
    [4,2],
    [0,2],
    [0,0]
]).T

# Horizontal shear
S = np.array([
    [1,0.8],
    [0,1]
])

sheared = S @ rect

plt.figure(figsize=(6,6))

plt.plot(rect[0], rect[1], lw=3, label="Original")
plt.plot(sheared[0], sheared[1], lw=3, label="Sheared")

plt.axis("equal")
plt.grid(True)
plt.xlim(-1,7)
plt.ylim(-1,4)
plt.legend()
plt.title("2D Shear")

plt.show()
```

# Rotation in 3d 

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

# Rectangle in the xy-plane
rect = np.array([
    [0,0,0],
    [4,0,0],
    [4,2,0],
    [0,2,0],
    [0,0,0]
]).T

# Rotation about z-axis
theta = np.radians(45)

R = np.array([
    [np.cos(theta),-np.sin(theta),0],
    [np.sin(theta), np.cos(theta),0],
    [0,0,1]
])

rotated = R @ rect

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111,projection='3d')

ax.plot(rect[0],rect[1],rect[2],
        lw=3,label='Original')

ax.plot(rotated[0],rotated[1],rotated[2],
        lw=3,label='Rotated')

ax.set_xlim(-3,5)
ax.set_ylim(-3,5)
ax.set_zlim(-1,3)

ax.legend()

plt.show()
```

# Reflection in 3d 

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

# -------------------------------------------------
# Rectangle in 3D
# -------------------------------------------------

rect = np.array([
    [0,0,1],
    [4,0,1],
    [4,2,1],
    [0,2,1],
    [0,0,1]
]).T

# -------------------------------------------------
# Reflection across the xy-plane
# -------------------------------------------------

R = np.array([
    [1,0,0],
    [0,1,0],
    [0,0,-1]
])

reflected = R @ rect

# -------------------------------------------------
# Plot
# -------------------------------------------------

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(rect[0], rect[1], rect[2],
        linewidth=3,
        label="Original")

ax.plot(reflected[0], reflected[1], reflected[2],
        linewidth=3,
        label="Reflected")

ax.set_xlim(-1,5)
ax.set_ylim(-1,3)
ax.set_zlim(-2,2)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.legend()
plt.title("Reflection across XY Plane")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D

# -------------------------------------------------
# Rectangle
# -------------------------------------------------

rect = np.array([
    [0,0,0],
    [4,0,0],
    [4,2,0],
    [0,2,0],
    [0,0,0]
]).T

# -------------------------------------------------
# Shear matrix
# x' = x + 0.8y
# -------------------------------------------------

S = np.array([
    [1,0.8,0],
    [0,1,0],
    [0,0,1]
])

sheared = S @ rect

# -------------------------------------------------
# Plot
# -------------------------------------------------

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(rect[0], rect[1], rect[2],
        linewidth=3,
        label="Original")

ax.plot(sheared[0], sheared[1], sheared[2],
        linewidth=3,
        label="Sheared")

ax.set_xlim(-1,7)
ax.set_ylim(-1,3)
ax.set_zlim(-1,2)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.legend()

plt.title("3D Shear")

plt.show()
```
