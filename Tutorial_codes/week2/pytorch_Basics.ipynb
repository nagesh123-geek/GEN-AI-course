```python
import time
import torch

# 1. Define the size of the square matrices for benchmarking
matrix_dimension = 10000

# 2. Initialize two large random matrices on the CPU
matrix_cpu_a = torch.randn(matrix_dimension, matrix_dimension)
matrix_cpu_b = torch.randn(matrix_dimension, matrix_dimension)

# 3. Measure matrix multiplication execution time on the CPU
start_time = time.time()
result_cpu = torch.matmul(matrix_cpu_a, matrix_cpu_b)
cpu_duration = time.time() - start_time

print(f"Time on CPU: {cpu_duration:.4f} seconds")

# 4. If CUDA is available, transfer matrices to GPU and measure performance
if torch.cuda.is_available():
    matrix_gpu_a = matrix_cpu_a.to('cuda')
    matrix_gpu_b = matrix_cpu_b.to('cuda')

    # Measure matrix multiplication execution time on the GPU
    start_time = time.time()
    result_gpu = torch.matmul(matrix_gpu_a, matrix_gpu_b)
    torch.cuda.synchronize()  # Wait for GPU computation to finish
    gpu_duration = time.time() - start_time

    print(f"Time on GPU: {gpu_duration:.4f} seconds")
    # 5. Calculate and display the speedup factor
    print("\nSpeedup (CPU time / GPU time):", cpu_duration / gpu_duration)
else:
    print("CUDA is not available on this system.")
```
