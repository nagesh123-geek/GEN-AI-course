```python
import numpy as np
import matplotlib.pyplot as plt

# Define a standard uniform 2D grid
x = np.linspace(-3, 3, 25)
y = np.linspace(-3, 3, 25)
X, Y = np.meshgrid(x, y)

# Activation functions
def relu(z):
    return np.maximum(0, z)

def leaky_relu(z, alpha=0.1):
    return np.where(z > 0, z, alpha * z)

def tanh(z):
    return np.tanh(z)

def gelu(z):
    return z * 0.5 * (
        1.0
        + np.tanh(
            np.sqrt(2 / np.pi)
            * (z + 0.044715 * z**3)
        )
    )

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(11, 11))
axes = axes.flatten()

titles = [
    "ReLU (The Space Folder)",
    "Leaky ReLU (The Asymmetric Bender)",
    "Tanh (The Space Compressor)",
    "GELU (The Smooth Hinge)"
]

activation_funcs = [relu, leaky_relu, tanh, gelu]

for ax, title, func in zip(axes, titles, activation_funcs):

    X_transformed = func(X)
    Y_transformed = func(Y)

    # Horizontal grid lines
    for i in range(X.shape[0]):
        ax.plot(
            X_transformed[i, :],
            Y_transformed[i, :],
            color="tab:blue",
            lw=1.2,
            alpha=0.7,
        )

    # Vertical grid lines
    for j in range(X.shape[1]):
        ax.plot(
            X_transformed[:, j],
            Y_transformed[:, j],
            color="tab:blue",
            lw=1.2,
            alpha=0.7,
        )

    ax.set_title(title, fontsize=14, weight="bold")
    ax.set_xlim(-3.5, 3.5)
    ax.set_ylim(-3.5, 3.5)
    ax.axhline(0, color="black", lw=1)
    ax.axvline(0, color="black", lw=1)
    ax.grid(True, linestyle="--", alpha=0.3)
    ax.set_aspect("equal")

plt.tight_layout()
plt.show()
```
